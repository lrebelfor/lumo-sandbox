```
# Cell 1 · setup (the same in every lab)
!git clone -q --depth 1 https://github.com/lrebelfor/lumo-sandbox /content/lumo 2>/dev/null || git -C /content/lumo pull -q
!pip install -q anthropic openai
import sys; sys.path.insert(0, "/content/lumo")
import lumo
from course_llm import LLM

ROUTE = "claude"   # your model route from Session 2: "claude", "openai", "gemini", "openrouter" (or "ollama" outside Colab)
llm = LLM(ROUTE)
print("Connected:", llm)
```

```
# Send your first message and see what it costs
r = llm.ask("In 3 sentences: what is an AI agent, and why do most never reach real users?")
print(r.text)
print("\nModel:", r.model, "| Tokens in:", r.tokens_in, "| Tokens out:", r.tokens_out, "| Cost:", r.cost)
```

```
# Try it on YOUR capstone task. Lumo track: a real Lumo email. Own case: replace both texts.
system_prompt = "You are an assistant for Lumo's support team, a small Dutch webshop for design lamps. Be friendly and concise."
task = "Draft a short reply to this customer email: " + lumo.emails()[0]["text"]

r = llm.ask(task, system=system_prompt)   # system = the standing instructions: role, tone, rules
print(r.text)
print("\nTokens in:", r.tokens_in, "| Tokens out:", r.tokens_out)
```

```
# Agent moment: a ready-made agent with one tool, on Lumo's order system. Just run it.
from lumo.tools import LOOKUP_ORDER
print(llm.run_agent("Hi, where is my order 1047? It's a birthday present.", tools=[LOOKUP_ORDER],
                    system="You are Lumo's support assistant. Check the order before answering. Be brief and kind."))
```
