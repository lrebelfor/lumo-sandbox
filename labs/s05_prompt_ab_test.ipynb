```
# Prompt A/B test: a weak and a strong prompt on the same email. Own case: swap in your own email and policy.
email = next(e["text"] for e in lumo.emails() if e["id"] == "E02")
policy = next(text for name, text in lumo.handbook() if name.startswith("06-")) + next(
    text for name, text in lumo.handbook() if name.startswith("03-"))

weak = f"Answer this email: {email}"

strong = f"""You are a support agent for Lumo, a small Dutch webshop for design lamps.

<policy>
{policy}
</policy>

<email>
{email}
</email>

Write the reply:
1. Apologise once, briefly.
2. Answer EVERY question in the email, using only the policy.
3. Ask for the order number and one photo of the damage.
Format: 4-6 sentences, friendly, no bullet points, sign as "Team Lumo"."""

for name, p in [("WEAK", weak), ("STRONG", strong)]:
    r = llm.ask(p)
    print(f"===== {name} ({r.tokens_in} tokens in) =====\n{r.text}\n")
```
