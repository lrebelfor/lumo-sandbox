```
!git clone -q --depth 1 https://github.com/lrebelfor/lumo-sandbox /content/lumo 2>/dev/null || git -C /content/lumo pull -q
!pip install -q chromadb
import sys; sys.path.insert(0, "/content/lumo")
import lumo, chromadb

# Chunks: one per paragraph of the Lumo handbook. Own case: your chunks from Session 7.
docs = []
for name, text in lumo.handbook():
    for n, para in enumerate(p for p in text.split("\n\n")[1:] if p.strip()):
        docs.append((f"{name[:-3]}-{n}", para.strip()))

db = chromadb.Client()                        # in-memory; chromadb.PersistentClient("db") saves to disk
kb = db.get_or_create_collection("handbook")  # Chroma turns each text into an embedding for you
kb.add(ids=[i for i, _ in docs], documents=[t for _, t in docs])
print(kb.count(), "chunks indexed. No API key needed today.")
```

```
def search(question, k=2):
    res = kb.query(query_texts=[question], n_results=k)
    for doc_id, text, dist in zip(res["ids"][0], res["documents"][0], res["distances"][0]):
        print(f"  {doc_id:18} distance {dist:.2f}  {text[:90]}")

# Semantic search finds meaning, not just matching words:
for q in ["How long until my parcel reaches Antwerp?",
          "My lamp came in pieces",
          "Can I pay with my phone wallet?"]:
    print(q); search(q); print()
```
