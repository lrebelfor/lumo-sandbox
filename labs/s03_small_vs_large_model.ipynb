```
# Cell 1 · setup (the same in every lab)
!git clone -q --depth 1 https://github.com/lrebelfor/lumo-sandbox /content/lumo 2>/dev/null || git -C /content/lumo pull -q
!pip install -q anthropic openai
import sys; sys.path.insert(0, "/content/lumo")
import lumo
from course_llm import LLM

ROUTE = "claude"   # your model route from Session 2: "claude", "openai", "gemini", "openrouter" (or "ollama" outside Colab)
llm = LLM(ROUTE)
print("Connected:", llm)
```

```
# Same tasks, two model sizes. Lumo track: three real support emails. Own case: write your own tasks.
emails = {e["id"]: e["text"] for e in lumo.emails()}
tasks = [
    "Summarise this customer email in one sentence: " + emails["E07"],
    "Classify the mood of this email (calm / annoyed / angry) and say what the customer wants: " + emails["E07"],
    "Write a polite 3-sentence reply to this email. Don't promise a refund: " + emails["E07"],
]

for model in [llm.model, llm.large_model]:
    print("=" * 20, model, "=" * 20)
    for t in tasks:
        r = llm.ask(t, model=model)
        cost = f"${r.cost:.5f}" if r.cost is not None else "look up the price for this model"
        print(f"\nTASK: {t[:70]}...\n{r.text}\n(tokens {r.tokens_in} in / {r.tokens_out} out, {cost})")
```
